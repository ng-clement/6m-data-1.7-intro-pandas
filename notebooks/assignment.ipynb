### Challenge 1: Build and Inspect the Dataset

Create the following DataFrame and perform a health check:

```python
import pandas as pd

books = pd.DataFrame({
    'title': ['The Pragmatic Programmer', 'Clean Code', 'Python Crash Course',
              'Data Science from Scratch', 'Designing Data-Intensive Apps'],
    'author': ['Hunt & Thomas', 'Robert Martin', 'Eric Matthes',
               'Joel Grus', 'Martin Kleppmann'],
    'genre': ['Tech', 'Tech', 'Tech', 'Data', 'Data'],
    'price': [45.99, 38.50, 29.99, 42.00, 55.00],
    'units_sold': [120, 95, 200, 80, 60],
    'in_stock': [True, True, False, True, False]
})
```

**Tasks:**
1. Use `.info()` — how many non-null values does each column have?
2. Use `.describe()` — what's the average price and maximum units sold?
3. Add a `revenue` column: `price × units_sold`.
4. What is the total revenue across all books?

````python
import pandas as pd
import numpy as np

books = pd.DataFrame({
    'title': ['The Pragmatic Programmer', 'Clean Code', 'Python Crash Course',
              'Data Science from Scratch', 'Designing Data-Intensive Apps'],
    'author': ['Hunt & Thomas', 'Robert Martin', 'Eric Matthes',
               'Joel Grus', 'Martin Kleppmann'],
    'genre': ['Tech', 'Tech', 'Tech', 'Data', 'Data'],
    'price': [45.99, 38.50, 29.99, 42.00, 55.00],
    'units_sold': [120, 95, 200, 80, 60],
    'in_stock': [True, True, False, True, False]
})
display(books)

print("1. Use `.info()` — how many non-null values does each column have?")
books.info()
print()

print("2. Use `.describe()` — what's the average price and maximum units sold?")
books.describe()
print(books.describe())
print()

print("3. Add a `revenue` column: `price × units_sold`.")
books['revenue'] = books['price'] * books['units_sold']
display(books)
print()

print("4. What is the total revenue across all books?")
total_revenue = books['revenue'].sum()
print(f"Total Revenue: ${total_revenue:,.2f}")
print()
````

### Challenge 2: Indexing & Filtering

**Tasks:**
1. Use `.loc` to select only `title` and `revenue` for Data genre books.
2. Use `.iloc` to retrieve the last 3 rows and first 3 columns.
3. Filter books that are **in stock** AND have a **price above £40**.
4. Find the book with the **highest revenue** using `.loc` with `.idxmax()`.


````python
display(books)

print("1. Use `.loc` to select only `title` and `revenue` for Data genre books.")
data_books = books.loc[books['genre'] == 'Data', ['title', 'revenue']]
display(data_books)

print("2. Use `.iloc` to retrieve the last 3 rows and first 3 columns.")
last_3_rows_first_3_columns = books.iloc[-3:,:3]
display(last_3_rows_first_3_columns)

print("3. Filter books that are **in stock** AND have a **price above £40**.")
in_stock_expensive = books.loc[(books['in_stock'] == True) & (books['price'] > 40)]
display(in_stock_expensive)

print("4. Find the book with the **highest revenue** using `.loc` with `.idxmax()`.")
highest_revenue_book = books.loc[books['revenue'].idxmax()]
display(highest_revenue_book)
````

### Challenge 3: Transform, Sort & Rank

**Tasks:**
1. Use `.apply()` to add a `price_tier` column: `'Budget'` (< £35), `'Standard'` (£35–£50), `'Premium'` (> £50).
2. Sort the DataFrame by `revenue` descending.
3. Add a `sales_rank` column using `.rank(ascending=False)`.
4. Use `.map()` to encode `genre` as a number: `{'Tech': 0, 'Data': 1}`.

````python
display(books)

print("1. Use `.apply()` to add a `price_tier` column: `'Budget'` (< £35), `'Standard'` (£35–£50), `'Premium'` (> £50).")
def price_tier(price):
    if price < 35:
        return 'Budget'
    elif 35 <= price <= 50:
        return 'Standard'
    else:
        return 'Premium'
books['price_tier'] = books['price'].apply(price_tier)
display(books)  

print("2. Sort the DataFrame by `revenue` descending.")
books_sorted = books.sort_values(by='revenue', ascending=False)
display(books_sorted)

print("3. Add a `sales_rank` column using `.rank(ascending=False)`.")
books['sales_rank'] = books['units_sold'].rank(ascending=False).astype(int)
display(books)

print("4. Use `.map()` to encode `genre` as a number: `{'Tech': 0, 'Data': 1}`.")
books['genre_encoded'] = books['genre'].map({'Tech': 0, 'Data': 1})
display(books)
````

Stretch Challenge (Optional)

The manager wants a report comparing average price and total revenue by `price_tier`. Use `groupby()` to generate this summary. Which tier contributes the most total revenue despite having the lowest average price?

````python
report = books.groupby('price_tier').agg({'price': 'mean', 'revenue': 'sum'}).reset_index()
report.rename(columns={'price': 'average_price', 'revenue': 'total_revenue'}, inplace=True)
display(report) 
````
